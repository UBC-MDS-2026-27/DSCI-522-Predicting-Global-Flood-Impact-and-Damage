```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

### Data Cleaning and Final Shape

```python
df = pd.read_excel("data/raw/Aggregated_Flood_Data.xlsx", sheet_name="Aggregated")

start = pd.to_datetime(df["Start Date"], errors="coerce")
end = pd.to_datetime(df["End Date"], errors="coerce")
df["Duration"] = (end - start).dt.days.clip(lower=0)

df = df[df["Year"].between(1900, 2026)].dropna(subset=["Fatalities"])

numeric_features = ["Year", "Lat", "Lon", "Duration", "Affected", "Injured"]
categorical_features = ["Country"]

X = df[numeric_features + categorical_features]
y = (df["Fatalities"] >= 1).astype(int)

X.shape, y.value_counts(normalize=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=522
)

# getting training data shape 

print("Shape of training set: ", X_train.shape)
```

### Balance Percentage dataframe of Target Variable

```python
# Getting Imbalance of Target (Fatal) Class

# creating a dataframe of our target y and associated counts with percentages
# to show balance vs imbalance

class_balance_df = pd.DataFrame({
    "Counts": y_train.value_counts(),
    "Percentag (%)": (y_train.value_counts(normalize = True) * 100).round(2),
})
class_balance_df
```

### Summary Table of all Numeric Features

```python
# Getting a Summary of all Numeric Features in the dataset

numeric_summary = pd.DataFrame({
    "Missing (%)": (X_train[numeric_features].isna().mean() * 100).round(2),
    "Min": X_train[numeric_features].min().round(2),
    "Median": X_train[numeric_features].median().round(2),
    "Mean": X_train[numeric_features].mean().round(2),
    "95th Percentile": X_train[numeric_features].quantile(0.95).round(2),
    "Max": X_train[numeric_features].max().round(2),
})
numeric_summary
```
