```python
#Imports
import pandas as pd
import openpyxl
```

```python
def data_download_xlsx_to_csv(url, save_dir, target_sheet, name):
    """Download .xlsx data from a target URL and save a sheet of it locally as a CSV in a target directory.
    
    Args:
    url : URL where the raw .xlsx file is stored
    save_dir : relative path where the downloaded data will be stored.
    target_sheet : Target sheet of the xlsx file.
    name : name for the downloaded file, no extension.
    
    """
    save_target = f"{save_dir}/{name}.csv"
    df = pd.read_excel(url, sheet_name=target_sheet)
    df.to_csv(save_target, index=False)
```

```python
data_url="https://www.sciencebase.gov/catalog/file/get/69e7e00db66b0164d0f7302d?f=__disk__c3%2F44%2Fd6%2Fc344d62cc5d115d408726fb52d9d459c6e1b5ed1"
data_dir = "data/raw"
target_sheet="Aggregated"
name="Aggregated_Flood_Data"

data_download_xlsx_to_csv(data_url, data_dir, target_sheet, name)
```
